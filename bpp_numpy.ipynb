```python
import numpy as np

arr = np.arange(1, 17).reshape(4, 4)

first_row = arr[0, :]
last_column = arr[:, -1]

print("Array:")
print(arr)

print("First row:")
print(first_row)

print("Last column:")
print(last_column)
```

```python
import numpy as np

a = np.array([1, 2, 3, 4])
b = np.array([5, 6, 7, 8])

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
```

```python
import numpy as np

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

dot_product = np.dot(a, b)

print(dot_product)
```

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

mean = np.mean(arr)
median = np.median(arr)
std = np.std(arr)

print("Mean:", mean)
print("Median:", median)
print("Standard deviation:", std)
```
