```python
set1 = {10, 20, 30, 40, 50}

num = int(input("Enter the item to remove: "))

if num in set1:
    set1.remove(num)
    print("Set after removing the item:", set1)
else:
    print("Item does not exist in the set")
```

```python
set1 = {10, 20, 30, 40}
set2 = {30, 40, 50, 60}

result = set1.intersection(set2)

print("Intersection of sets:", result)
```

```python
set1 = {10, 20, 30, 40}
set2 = {30, 40, 50, 60}

result = set1.union(set2)

print("Union of sets:", result)
```

```python
set1 = {10, 25, 5, 40, 15}

print("Maximum value:", max(set1))
print("Minimum value:", min(set1))
```
