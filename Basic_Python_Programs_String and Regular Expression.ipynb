```python
string = input("Enter a string: ")

upper = 0
lower = 0

for ch in string:
    if ch.isupper():
        upper += 1
    elif ch.islower():
        lower += 1

print("Number of uppercase letters:", upper)
print("Number of lowercase letters:", lower)
```

```python
string = input("Enter a string: ")

if string == string[::-1]:
    print("The string is a palindrome")
else:
    print("The string is not a palindrome")
```

```python
string = input("Enter a string: ")

n = len(string)

result = string[:2] * n

print("Output:", result)
```

```python
string = input("Enter a string: ")

if string.startswith('x'):
    string = string[1:]

if string.endswith('x'):
    string = string[:-1]

print("Output:", string)
```

```python
string = input("Enter a string: ")
n = int(input("Enter n: "))

last_chars = string[-n:]

result = last_chars * n

print("Output:", result)
```

```python
import re

string = input("Enter a string: ")

pattern = r"^[bh][aiu]t$"

if re.match(pattern, string):
    print("String matched")
else:
    print("String not matched")
```

```python
import re

string = input("Enter two words: ")

pattern = r"^[A-Za-z]+ [A-Za-z]+$"

if re.match(pattern, string):
    print("String matched")
else:
    print("String not matched")
```

```python
import re

string = input("Enter last name and first initial: ")

pattern = r"^[A-Za-z]+, [A-Za-z]$"

if re.match(pattern, string):
    print("String matched")
else:
    print("String not matched")
```
