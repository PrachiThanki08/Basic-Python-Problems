```python
t = (10, 20, 30, 40, 50, 60, 70)

print("4th element from first:", t[3])
print("4th element from last:", t[-4])
```

```python
t = (10, 20, 30, 40, 50)

num = int(input("Enter an element: "))

if num in t:
    print("Element exists in the tuple")
else:
    print("Element does not exist in the tuple")
```

```python
list1 = [10, 20, 30, 40, 50]

t = tuple(list1)

print("List:", list1)
print("Tuple:", t)
```

```python
t = (10, 20, 30, 40, 50)

num = int(input("Enter an element: "))

if num in t:
    print("Index of", num, "is", t.index(num))
else:
    print("Element does not exist in the tuple")
```
